# Task 6

```python
# Scatter Plot

plt.figure(figsize=(14,6))
plt.scatter(df['likes'],df['impression_count'],alpha=0.3,s=15)

plt.xlabel("Likes")
plt.ylabel("Impressions")
plt.title("Likes vs Impressions")
plt.grid(True)
plt.show()
```

```python
# Line Plot 

daily_engagement = df.groupby('posted_at')['engagement_score'].mean()

plt.figure(figsize=(14,6))
plt.plot(
    daily_engagement.index,
    daily_engagement.values,
    color='black',
    marker='o',
    markerfacecolor='red',
    markeredgecolor='red',
    markersize=3,
    linewidth=1
)
plt.xlabel("Date")
plt.ylabel("Average engagement score")
plt.title("Daily engagement trend")
plt.grid(alpha=0.3)
plt.xticks(rotation=45)

plt.show()
```

```python
# Bar Plot

category_counts = df['post_category'].value_counts()
plt.figure(figsize=(14, 6))

plt.bar(
    category_counts.index,
    category_counts.values,
    color='green'
)
plt.xlabel("Post category")
plt.ylabel("Number of posts")
plt.title("Posts by category")

plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Pie Plot

gender_counts = df['gender'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Gender Distribution")

plt.show()
```

```python
# Histogram

plt.figure(figsize=(10, 6))
plt.hist(
    df['age'],
    bins=10,
    color='orange',
    edgecolor='black'
)

plt.xlabel("Age")
plt.ylabel("Number of Users")
plt.title("Age Distribution")
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Box plot

plt.figure(figsize=(8, 6))
plt.boxplot(df['engagement_rate'])
plt.ylabel("Engagement Rate")
plt.title("Engagement Rate Distribution")
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Count Plot

plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x='post_type'
)

plt.xlabel("Post Type")
plt.ylabel("Number of Posts")
plt.title("Number of Posts by Post Type")

plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Bar Plot

plt.figure(figsize=(10, 6))
sns.barplot(
    data=df,
    x='post_category',
    y='likes',
    color='yellow'
)
plt.xlabel("Post Category")
plt.ylabel("Average Likes")
plt.title("Average Likes by Post Category")
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Violin Plot

plt.figure(figsize=(9, 6))
sns.violinplot(
    data=df,
    x='sentiment',
    y='follower_count'
)
plt.xlabel("Sentiment")
plt.ylabel("Follower count")
plt.title("Follower count by sentiment")
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Pair Plot

sample_df = df[['likes', 'comments', 'shares', 'watch_time_sec', 'engagement_rate']].sample(1000, random_state=42)
sns.pairplot(sample_df)

plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='YlGnBu'
)

plt.title("Correlation Matrix")

plt.show()
```

```python
# Swarn Plot

swarm_df = df.sample(500, random_state=42)
plt.figure(figsize=(10, 6))
sns.swarmplot(
    data=swarm_df,
    x='device_type',
    y='engagement_score')
plt.xlabel("Device Type")
plt.ylabel("Engagement Score")
plt.title("Engagement Score by Device Type")
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# Interactive line plot

daily_engagement_df = df.groupby('posted_at',as_index=False)['engagement_score'].mean()
fig = px.line(
    daily_engagement_df,
    x='posted_at',
    y='engagement_score',
    title='Daily Engagement Trend',
    markers=True
)
fig.update_layout(
    xaxis_title='Date',
    yaxis_title='Average Engagement Score'
)

fig.show()
```

```python
avg_engagement = df.groupby(
    'post_type',
    as_index=False
)['engagement_score'].mean()

fig = px.bar(
    avg_engagement,
    x='post_type',
    y='engagement_score',
    title='Average Engagement Score by Post Type',
    text_auto='.2f'
)

fig.update_layout(
    xaxis_title='Post type',
    yaxis_title='Average engagement score'
)

fig.show()
```

## Which post types have highest engagement? 

```python
post_type_engagement = df.groupby('post_type')['engagement_score'].mean().sort_values(ascending=False)
post_type_engagement
```

## Best-performing content category? 

```python
category_engagement = df.groupby('post_category')['engagement_score'].mean().sort_values(ascending=False)
category_engagement
```

## Which countries have highest average engagement rate? 

```python
country_engagement_rate= df.groupby('country')['engagement_rate'].mean().sort_values(ascending=False)
country_engagement_rate
```

## How age affects engagement.

```python
df['age_group'] = pd.cut(df['age'],
    bins=[12, 18, 25, 35, 45, 55, 65],
    labels=['13-18', '19-25', '26-35', '36-45', '46-55', '56-64']
)
age_engagement = df.groupby('age_group',observed=True)['engagement_score'].mean()

age_engagement
```

## Performance difference for verified accounts.

```python
verified_engagement = df.groupby('is_verified')['engagement_score'].mean()
verified_engagement
```

## Behavioral Insights

```python
device_watch_time = df.groupby('device_type')['watch_time_sec'].mean().sort_values(ascending=False)
device_watch_time
```
